```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.df = pd.read_csv("../data/weather.csv")

df["timestamp"] = pd.to_datetime(
    df[["year", "month", "day", "hour"]]
)

df = df.sort_values("timestamp").reset_index(drop=True)
```

```python
df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24)
df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24)

df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)

df["day_of_year"] = df["timestamp"].dt.dayofyear
```

```python
df["wdir_sin"] = np.sin(
    np.deg2rad(df["wdir"])
)

df["wdir_cos"] = np.cos(
    np.deg2rad(df["wdir"])
)
```

```python
weather_variables = [
    "temp",
    "rhum",
    "prcp",
    "wspd",
    "pres"
]

for col in weather_variables:
    df[f"{col}_diff"] = df[col].diff()
```

```python
for col in ["temp", "rhum", "wspd", "pres"]:
    
    df[f"{col}_rolling_mean_24h"] = (
        df[col]
        .shift(1)
        .rolling(window=24, min_periods=12)
        .mean()
    )
    
    df[f"{col}_rolling_std_24h"] = (
        df[col]
        .shift(1)
        .rolling(window=24, min_periods=12)
        .std()
    )
```

```python
for col in ["temp", "rhum", "wspd", "pres"]:
    
    df[f"{col}_deviation"] = (
        df[col] -
        df[f"{col}_rolling_mean_24h"]
    )
```

```python
df.to_csv(
    "../data/engineered_weather.csv",
    index=False
)

print("Engineered dataset saved.")
print("Shape:", df.shape)
```
